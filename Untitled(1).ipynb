```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Prepare the data
x = df1[['x']]
y = df1['y']
```

```python
# Split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Create a linear regression model
model = LinearRegression()
```

```python
# Train the model
model.fit(x_train, y_train)
```

```python
# Test the model
y_pred = model.predict(x_test)
```

```python
# Print results
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("R-squared:", r2_score(y_test, y_pred))
```

```python
# Values to predict
values_to_predict = pd.DataFrame({'x': [120, 126, 132]})
```

```python
# Predictions
predictions = model.predict(values_to_predict)
```

```python
print("Predicted Results for x=120, 126, 132:", predictions)
```

```python
# Plotting predictions
plt.figure(figsize=(10, 6))

plt.scatter(df1['x'], df1['y'], color='blue', label='train data')
plt.scatter(values_to_predict['x'], predictions, color='red', label='Predictions')
plt.plot(df1['x'], model.predict(df1[['x']]), color='green', label='Regression Line')

plt.title('Linear Regression Model Predictions')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.legend()

plt.show()
```
